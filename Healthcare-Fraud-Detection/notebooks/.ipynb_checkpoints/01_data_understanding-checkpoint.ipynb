```python
import pandas as pd
fraud = pd.read_csv("D:\Project\Healthcare-Fraud-Detection\data\processed\Train-1542865627584.csv")
beneficiary = pd.read_csv("D:\Project\Healthcare-Fraud-Detection\data\processed\Train_Beneficiarydata-1542865627584.csv")
inpatient = pd.read_csv("D:\Project\Healthcare-Fraud-Detection\data\processed\Train_Inpatientdata-1542865627584.csv")
outpatient = pd.read_csv("D:\Project\Healthcare-Fraud-Detection\data\processed\Train_Outpatientdata-1542865627584.csv")
```

```python
print(fraud.shape)
print(beneficiary.shape)
print(inpatient.shape)
print(outpatient.shape)
```

```python
fraud["PotentialFraud"] = fraud["PotentialFraud"].map({"Yes":1, "No":0})
```

```python
inpatient = inpatient.merge(beneficiary, on='BeneID', how='left')
outpatient = outpatient.merge(beneficiary, on='BeneID', how='left')
```

```python
outpatient.columns
```

```python
all_claim=pd.concat([inpatient,outpatient])
```

```python
df=all_claim.merge(fraud, on='Provider', how='left')
```

```python
# print(f"Total Claims: {final_df.shape[0]}")
df.columns
```

```python
df.drop(columns=['BeneID', 'ClaimID','AttendingPhysician', 'OperatingPhysician',
       'OtherPhysician'],axis=1,inplace=True)
```

```python
df.columns
```

```python
df['ClaimStartDt'] = pd.to_datetime(df['ClaimStartDt'])
df['ClaimEndDt'] = pd.to_datetime(df['ClaimEndDt'])
df['ClaimDuration'] = (df['ClaimEndDt'] - df['ClaimStartDt']).dt.days
```

```python
df.isnull().sum()
```

```python
df['AdmissionDt'] = pd.to_datetime(df['AdmissionDt'])
df['DischargeDt'] = pd.to_datetime(df['DischargeDt'])
df['StayDuration'] = (df['DischargeDt'] - df['AdmissionDt']).dt.days
df['AdmissionDt'].isnull().sum()
```

```python
df['DOB'] = pd.to_datetime(df['DOB'])
df['Age'] = 2026- df['DOB'].dt.year
```

```python
df['DOB']
```

```python
df.drop(columns=['ClaimStartDt','ClaimEndDt','AdmissionDt','DischargeDt','DOB','DOD'],axis=1,inplace=True)
```

```python
df.columns
```

```python
df.drop('Provider', axis=1, inplace=True)
```

```python
diag_cols = [col for col in df.columns if 'ClmDiagnosisCode' in col]
df['Num_Diagnosis'] = df[diag_cols].notnull().sum(axis=1)
df.drop(diag_cols, axis=1, inplace=True)
df.drop(['ClmAdmitDiagnosisCode', 'DiagnosisGroupCode'], axis=1, inplace=True)
```

```python
proc_cols = [col for col in df.columns if 'ClmProcedureCode' in col]
df['Num_Procedures'] = df[proc_cols].notnull().sum(axis=1)
df.drop(proc_cols, axis=1, inplace=True)
```

```python
df.drop('County', axis=1, inplace=True)
```

```python
df.columns
```
